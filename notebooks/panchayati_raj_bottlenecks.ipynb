# Where Panchayati Raj grievances get stuck

This note has one purpose: to find where grievances against the **Panchayati Raj
& Drinking Water** department lose time, and to say which of those places an
intervention could plausibly move.

It is in two parts.

**Part 1 describes the department as a whole** — how many grievances arrive, by
what route, about what, and how long they take.

**Part 2 narrows to a single subcategory**, rural housing. Complaint complexity
varies by type, so comparing two offices across the whole department confounds
how an office performs with what happens to reach it. Holding the subcategory
fixed removes the part of that variation we can observe. Section 1.8 shows how
large it is.

Rural housing is the department's largest subcategory by a wide margin, and the
scheme-named housing subcategory (IAY / MKY / BPGY / PMAY) is the second. They
are the same policy area recorded under two labels; this note follows the larger
of them and section 1.8 shows both.

The same analysis for SSEPD is in `ssepd_bottlenecks.ipynb`. Both notebooks run
the same code — `janasunani.analytics.bottlenecks` — over a different
department, so a number that differs between them is a difference in the
department, not in the method.

**Panchayati Raj is by far the largest department in the corpus**: about half a
million filings over four years, against SSEPD's forty thousand. Nothing in the
method changes, but every count here is an order of magnitude larger and the
small-sample caveats that bind the SSEPD note mostly do not bind this one.

### Reading this note

**The sample narrows as the note goes on, and never widens.** Each part opens
with its own funnel, every exhibit is stamped with the step it stands on, and
that step's count is printed on the exhibit. Where a section needs a narrower
base than the one before it — resolution *time* needs disposed grievances,
resolution *status* does not — the two are separate exhibits rather than one
that quietly changes base.

**Every duration is reported as a mean and a median.** They answer different
questions here: the mean of time to close runs well above the median, because a
minority of grievances run very long. Distributions are shown for 2024-25.

**Decompositions use means only.** The mean of a total is the sum of the means
of its parts; the median of a total is not the sum of the parts' medians. Only
the mean can be split into phases and add back up.

**Counts carry their share in parentheses** wherever a share is meaningful.

Years run **July to June**. No internal office codes, no officer names, and no
citizen's own words appear anywhere.

## 1.2 The same complaint, filed more than once

A citizen chasing an unresolved case files again. Counting each filing as a
separate grievance overstates how many people are waiting and understates how
long each of them waits.

Filings are collapsed when they are **the same complaint from the same person**.
Both halves must hold: candidate pairs come from a shared petitioner identity (a
salted hash of a mobile number or email, never the number itself), and each pair
is then checked against the text of the two filings, collapsing only if the
redacted texts are at least half identical. This is the repository's own
duplicate index, not a rule invented for this note.

Grouping ran **once over all four years**, not year by year: the identity side
of the match is deliberately not blocked by district or time window, so a
resubmission months later and elsewhere is still found.

## 2.5 How the time is measured, before it is cut three ways

Sections 2.6 to 2.8 ask the same question of the same claims three times over:
for the subcategory as a whole, by where the claim entered, and by the route it
took. Each shows the total time to close and the same split of it. So the split
is defined once, here, before any exhibit stands on it.

Total time to close divides into five consecutive spans that tile:

| Phase | Runs from | Runs to |
|---|---|---|
| **Registration** | the claim is filed | the first action anyone records |
| **First assignment** | that first action | it is passed to a different office for the first time |
| **Field action** | that first hand-off | it is sent back up to a higher office |
| **Review** | it is sent back up | the last action recorded on it |
| **Closure** | that last action | it is formally recorded as resolved |

**The decomposition reports means, not medians, and that is not a stylistic
choice.** The mean of a total equals the sum of the means of its parts, so a
mean can be split and the pieces add back to the whole. Medians do not add: the
middle claim for one phase is a different claim from the middle for another, and
five phase medians will not sum to the median total. Medians appear beside the
means in the tables, but only the means are decomposed.

A route is the sequence of offices that held the claim, named by **role and
never by place**: Collector, BDO, Secretary, and so on, but never "Collector,
Khordha". Consecutive spells in the same role collapse, so a claim worked twice
by one Collector is one step. An office whose code the rung ladder can place but
not name appears by its level, as "Other district office" and the like, rather
than dropping the claim from the route for want of a label.

This section opens the last two steps of the funnel. **P3** needs dated action
history in which every office is one the ladder recognises. **P4** needs the
five spans to sum to the total.

## What these numbers can and cannot say

**Nothing here measures whether a decision was right.** "Disposed" records that a
file was closed. Only "disposed with benefit" records that something reached the
citizen, and neither says whether the person was entitled to it.

**A long wait is not idleness.** Time between recorded steps may be field
enquiry, a statutory waiting period, or time spent waiting on the citizen. No
figure here should be read as time wasted, and none of it identifies an
individual officer — the record's officer-name column is never read.

**Holding the subcategory fixed removes only observed complexity.** Two
rural housing claims can still differ in ways the record does not show. Part 2 removes the
variation between subcategories, not the variation within one.

**Rural housing and IAY/MKY/BPGY/PMAY are the same policy area under two
labels.** Part 2 follows the larger of them. A finding here is about how rural
housing grievances recorded under that label travel, not about the housing
programme as a whole.

**Repeat filings are collapsed across all four years, but only within
Panchayati Raj.** A claim filed once here and once against another department is
counted twice. The threshold is the duplicate index's, not tuned here: it will
miss a resubmission written afresh in different words, and will collapse two
genuinely separate claims a person described identically. Section 1.2 reports
how much of the collapse comes from very large identity groups, which look like
shared intake devices rather than individuals.

**Open claims are excluded from every timing figure**, so recent periods look
faster than they will prove.

**Routing coverage is not complete**, and the phase decomposition drops a
further tranche whose last action postdates its closing date. Both are counted
where they happen rather than absorbed silently.
